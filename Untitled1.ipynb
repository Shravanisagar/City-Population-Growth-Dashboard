```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.DataFrame({
    'comment': [
        "What a fantastic goal!",
        "India loses another wicket.",
        "The player is ready for the match.",
        "What a thrilling finish!",
        "Rain has stopped the play.",
        "The referee blows the whistle.",
        "Unbelievable comeback!",
        "Poor performance today.",
        "The match begins slowly."
    ],
    'emotion': [
        "Excited",
        "Sad",
        "Neutral",
        "Excited",
        "Sad",
        "Neutral",
        "Excited",
        "Sad",
        "Neutral"
    ]
})

data.head()
```

```python
X = data['comment']
y = data['emotion']
```

```python
vectorizer = TfidfVectorizer()
X_vec = vectorizer.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_vec, y, test_size=0.3, random_state=42)

```

```python
model = RandomForestClassifier()
model.fit(X_train, y_train)
print("✅ Model training complete!")
```

```python
# Predictions
y_pred = model.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)
print("✅ Accuracy:", acc)

# Confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)

# Plot the graph
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=model.classes_,
            yticklabels=model.classes_)
plt.title("Emotion Detection Confusion Matrix")
plt.xlabel("Predicted Emotion")
plt.ylabel("True Emotion")
plt.show()

```

```python
comment = input("Enter a sports commentary: ")

vec = vectorizer.transform([comment])
prediction = model.predict(vec)[0]

print("🎯 Detected Emotion:", prediction)
```
